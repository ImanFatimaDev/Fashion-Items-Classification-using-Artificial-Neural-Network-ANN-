# 1: Libraries Import

This cell imports the necessary Python libraries required for the project. TensorFlow and Keras are used to create, train, and evaluate the Artificial Neural Network (ANN). Matplotlib is used to visualize data, graphs, and model performance throughout the project.

```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import fashion_mnist
```

# 2: Dataset Load

In this cell, the Fashion-MNIST dataset is loaded directly from Keras. The dataset is automatically divided into training and testing sets, where the training set is used to train the ANN and the test set is used to evaluate its performance on unseen data.

```python
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
```

# 3: Data Normalization

This cell normalizes the pixel values of the images from the original range of 0–255 to a range between 0 and 1. Normalization helps the model learn faster and improves overall training stability and performance.

```python
train_images = train_images / 255.0
test_images = test_images / 255.0
```

# 4: ANN Model Creation

Here, a simple Artificial Neural Network is created using the Keras Sequential API. The model consists of an input layer that flattens the images, one hidden layer with ReLU activation, and an output layer with softmax activation for predicting the 10 classes of fashion items.

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
```

# 5: Model Compilation

This cell compiles the ANN model by specifying the optimizer, loss function, and evaluation metrics. The Adam optimizer is used for efficient training, sparse categorical cross-entropy is used for multi-class classification, and accuracy is tracked as the performance metric.

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

# 6: Model Training

In this cell, the ANN model is trained on the normalized training dataset for a fixed number of epochs. During training, the model adjusts its weights to minimize the loss function and improve its ability to correctly classify fashion items.

```python
model.fit(train_images, train_labels, epochs=5)
```

# 7: Model Evaluation

This cell evaluates the trained model using the test dataset, which was not used during training. It prints the test accuracy to show how well the ANN generalizes to new, unseen images.

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print("Test Accuracy:", test_acc)
```

# 8: Accuracy Graph (Training Performance)

This cell retrains the model using a validation split to monitor both training and validation performance during the learning process. This helps detect overfitting and ensures that the model performs well on unseen data.

```python
history = model.fit(train_images, train_labels, epochs=5, validation_split=0.1)
```

# 9: Plot Accuracy Graph

This cell plots the training and validation accuracy across epochs using Matplotlib. The graph provides a visual understanding of how the model’s performance improves over time and whether it is learning effectively.

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Model Accuracy')
plt.legend(['Training Accuracy', 'Validation Accuracy'])
plt.show()
```

# 10: Final Evaluation

The final evaluation cell tests the model on the test dataset one more time and prints the final accuracy. This confirms the overall performance of the ANN on unseen data and provides a metric for the success of the project.

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print("Final Test Accuracy:", test_acc)
```

# Results

The trained model successfully classified fashion items and achieved good accuracy on the test dataset, demonstrating the effectiveness of the Artificial Neural Network.
